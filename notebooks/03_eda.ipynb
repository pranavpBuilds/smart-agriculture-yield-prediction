# 03 - Exploratory Data Analysis

```python
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('../data/raw/yield_df.csv').drop(columns=['Unnamed: 0'])
sns.set_theme(style='whitegrid')
```

## Yield distribution

```python
plt.figure(figsize=(8,4))
sns.histplot(df['hg/ha_yield'], bins=50, color='#2f8f4e')
plt.title('Yield Distribution (hg/ha)')
plt.show()
```

## Average yield by crop

```python
df.groupby('Item')['hg/ha_yield'].mean().sort_values(ascending=False)
```

## Average yield by area (top 10)

```python
df.groupby('Area')['hg/ha_yield'].mean().sort_values(ascending=False).head(10)
```

## Yield trend over years

```python
trend = df.groupby('Year')['hg/ha_yield'].mean()
plt.figure(figsize=(8,4))
trend.plot(marker='o', color='#2f8f4e')
plt.title('Average Yield by Year')
plt.show()
```

## Rainfall / Temperature / Pesticides vs Yield

```python
fig, axes = plt.subplots(1, 3, figsize=(15,4))
sns.scatterplot(data=df.sample(1000, random_state=42), x='average_rain_fall_mm_per_year', y='hg/ha_yield', ax=axes[0], color='#2f8f4e', alpha=0.5)
sns.scatterplot(data=df.sample(1000, random_state=42), x='avg_temp', y='hg/ha_yield', ax=axes[1], color='#b8860b', alpha=0.5)
sns.scatterplot(data=df.sample(1000, random_state=42), x='pesticides_tonnes', y='hg/ha_yield', ax=axes[2], color='#5b6b60', alpha=0.5)
plt.tight_layout()
plt.show()
```

## Correlation heatmap

```python
corr_cols = ['hg/ha_yield','average_rain_fall_mm_per_year','pesticides_tonnes','avg_temp','Year']
plt.figure(figsize=(6,5))
sns.heatmap(df[corr_cols].corr(), annot=True, cmap='Greens')
plt.show()
```

## Top / bottom crops and areas by average yield

```python
means_crop = df.groupby('Item')['hg/ha_yield'].mean().sort_values(ascending=False)
print('Top crops:\n', means_crop.head(3))
print('\nBottom crops:\n', means_crop.tail(3))
```

```python
means_area = df.groupby('Area')['hg/ha_yield'].mean().sort_values(ascending=False)
print('Top areas:\n', means_area.head(3))
print('\nBottom areas:\n', means_area.tail(3))
```
