# 05 - Model Comparison
Reads the metrics produced by `train_models.py` (`reports/metrics.json`) and visualizes the comparison — no numbers are re-entered by hand.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt

with open('../reports/metrics.json') as f:
    report = json.load(f)
metrics = report['metrics']
best = report['best_model']
print('Best model:', best)
```

```python
comparison = pd.DataFrame({m: metrics[m]['test'] for m in metrics}).T
comparison
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14,4))
comparison['MAE'].plot(kind='bar', ax=axes[0], color='#2f8f4e', title='MAE')
comparison['RMSE'].plot(kind='bar', ax=axes[1], color='#4cae6a', title='RMSE')
comparison['R2'].plot(kind='bar', ax=axes[2], color='#1f6b38', title='R2')
plt.tight_layout()
plt.show()
```

The model with the lowest test RMSE (`see report['best_model']`) is selected as the production model used by the FastAPI backend and saved as `models/<best>_pipeline.joblib`, referenced by `models/best_model.json`.
